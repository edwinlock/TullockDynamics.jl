# TullockDynamics.jl

## Tutorial
A brief and incomplete introduction to the `TullockDynamics.jl` package I created. First we need to load the package and install some additional packages.

```julia
import Pkg; Pkg.add("Measures")
```

```julia
using Revise
using Plots
using Measures

# Load TullockDynamics package
using Pkg; Pkg.activate("../TullockDynamics.jl")
Pkg.instantiate()
using TullockDynamics
```

### Set up agents

I'll first describe how to define a general agent by setting her parameters appropriately. It's possible to create contests with different kinds of agents, who have heterogeneous parameters.

```julia
linear_cost(x) = x  # agent has linear cost function
p(t) = 1  # agent plays in every round
window(t) = 1:t-1  # agent has access to full history
α(t) = 1  # agent commits fully to best response
χ = 0.05  # lower bound on agent's effort
estimator = max_likelihood_estimator;  # agent follows MLE mechanism
```

Now we can create an agent, or even a vector of agents, with these parameters. For simplicity, I'll create a vector of `n` homogeneous agents who all share the same parameters.

```julia
n = 3  # number of agents
a = Agent(linear_cost, p, window, α, estimator, χ)  # a single agents
agents = [Agent(linear_cost, p, window, α, estimator, χ) for _ in 1:n]  # a vector of agents
initial_efforts = [rand() for _ in 1:n];
```

My package also provides convenience constructors `MLEAgent`, `DetMLEAgent` and `DumbAgent` that set the above defaults for `p`, `window` and `α`, and choose the appropriate estimator. So we could have constructed the agents by writing:

```julia
# agents = [DetMLEAgent(linear_cost; δ = δ) for _ in 1:n];
agents = [MLEAgent(linear_cost; δ = δ) for _ in 1:n];
# agents = [DumbAgent(linear_cost; δ = δ) for _ in 1:n];
```

### Creating a Tullock contest
Now we create the contest using the agents defined above, and fixing the number of rounds `T`.

```julia
T = 2000
contest = TullockContest(agents, initial_efforts, T);
```

### Running and evaluating the dynamics
Now we can run the dynamics.

```julia
run!(contest)
```

The `contest` object stores the progress of the Tullock contest in four matrices:
* `contest.efforts`
* `contest.utilities`
* `contest.nash_gaps`
* `contest.winners`

For any such matrix, the entry `matrix[i,t]` is the value for agent `i` in round `t`. We can, e.g., see the efforts of all the agents in round `t` as follows:

```julia
t = 3
contest.efforts[:,t]
```

### Plotting the output

The package provides a convenience function to plot the efforts, utilities and Nash gaps. Optionally, you can specify the y-axis range as a keyword argument `ylims=(min, max)`.

```julia
visualise(contest, rounds=(10,T))
```

We can also generate individual plots, e.g., the efforts in given rounds:

```julia
efforts_plt = visualise(contest.efforts, rounds=(10,T), ylabel="effort")
plot(efforts_plt, size=(1000,400), margin=10mm)
```

## Experiments

### Classic Dynamics with Homogeneous Agents

We'll create `n` homogeneous agents with classic dynamics. Each agent sees their opponents' total efforts in each round, and best responds to that.

```julia
# Set agent parameters
n = 20  # number of agents
p(t) = 1  # agent plays in every round
window(t) = 1:t-1  # agent has access to full history
α(t) = 0.1  # agent commits fully to best response
δ = 0.01  # lower bound on agent's effort
estimator = classic_estimator;  # agent follows classic dynamics

# Create agents and initial efforts
agents = [Agent(linear_cost, p, window, α, estimator, δ) for _ in 1:n]
initial_efforts = [0.25*rand() for _ in 1:n]
```

```julia
# Create contest
T = 100  # number of rounds
contest = TullockContest(agents, initial_efforts, T);
```

```julia
run!(contest)
```

```julia
visualise(contest, rounds=(1, T))
```
